```python
import pandas as pd
pd.set_option('display.max_columns', 100)
pd.set_option('display.max_rows', 50)
```

```python
sales_df = pd.read_csv("/Users/kushal09/Documents/Documents - Krrish's MacBook Air/Career247 /Capstone_project/inventory_optimization/data/raw/sales_fact.csv")
products_df = pd.read_csv("/Users/kushal09/Documents/Documents - Krrish's MacBook Air/Career247 /Capstone_project/inventory_optimization/data/raw/products_master.csv")
suppliers_df = pd.read_csv("/Users/kushal09/Documents/Documents - Krrish's MacBook Air/Career247 /Capstone_project/inventory_optimization/data/raw/suppliers_master.csv")
inventory_df = pd.read_csv("/Users/kushal09/Documents/Documents - Krrish's MacBook Air/Career247 /Capstone_project/inventory_optimization/data/raw/inventory_snapshot.csv")
```

```python
sales_product = sales_df.merge(products_df , on = 'sku_id' , how= 'left')

sales_product_supplier = sales_product.merge(suppliers_df , on = 'supplier_id' , how = 'left')

final_df = sales_product_supplier.merge(inventory_df , on = 'sku_id' , how = 'left')
```

```python
print(final_df.columns[final_df.columns.duplicated()]) # After mergeing here we are checking if any duplicate column get created or not.
```

```python
final_df.shape
```

```python
# Here we are observing that the data has the columns with suffix '_x' and '_y'
final_df.columns
```

```python
rename_x = {
    col: col.replace("_x", "")
    for col in final_df.columns
    if "_x" in col
}   #  In this line of code I am definig the conditoin in the object  removing the _x from suffix 

rename_y = {
    col: col.replace("_y", "")
    for col in final_df.columns
    if "_y" in col
} #  In this line of code I am definig the conditoin in the object  removing the _y from suffix 

final_df = final_df.rename(columns=rename_x)
final_df = final_df.rename(columns=rename_y)
```
