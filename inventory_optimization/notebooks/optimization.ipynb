```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns' , None)
```

```python
sales = pd.read_csv("inventory_optimization/data/processed/sales_clean.csv")

products = pd.read_csv("inventory_optimization/data/processed/products_clean.csv")

inventory = pd.read_csv("inventory_optimization/data/processed/inventory_clean.csv")

suppliers = pd.read_csv("inventory_optimization/data/processed/suppliers_clean.csv")
```

```python
print(sales.shape)
print(products.shape)
print(inventory.shape)
print(suppliers.shape)
```

```python
print(sales.columns)

print(products.columns)

print(inventory.columns)

print(suppliers.columns)
```

```python
master_df = sales.merge(
    products,
    on="sku_id",
    how="left"
)
```

```python
master_df = master_df.merge(
    inventory,
    on="sku_id",
    how="left"
)
```

```python
master_df = master_df.merge(
    suppliers,
    on="supplier_id",
    how="left"
)
```
