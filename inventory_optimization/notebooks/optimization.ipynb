```python
sa_pro = sales.merge(
    products,
    on="sku_id",
    how="left",
    validate="many_to_one"
)

sa_pro_sup = sa_pro.merge(
    suppliers,
    on="supplier_id",
    how="left",
    validate="many_to_one"
)

master_df = sa_pro_sup.merge(
    inventory,
    on="sku_id",
    how="left",
    validate="many_to_one"
)
```
